# 컴퓨터 비전


컴퓨터 비전은 인공지능(AI)의 한 분야로, 컴퓨터가 인간처럼 이미지나 영상을 이해하고 분석할 수 있도록 하는 기술입니다. 객체 검출, 이미지 분류, 얼굴 인식, 장면 이해 등 다양한 작업을 포함하며, 주로 머신러닝과 딥러닝을 활용해 이미지 속 특징을 추출하고 패턴을 학습합니다.

예를 들어 자율주행 자동차는 컴퓨터 비전을 사용하여 도로의 차선, 보행자, 신호등을 인식합니다. 의료 영상 분석에서는 CT나 MRI 이미지를 분석하여 질병 진단을 돕습니다. 최근에는 합성곱 신경망(CNN)과 같은 딥러닝 모델의 발전으로 컴퓨터 비전 성능이 크게 향상되었으며, 다양한 산업에서 활발히 활용되고 있습니다.

## 1. 컴퓨터 비전 프레임워크

컴퓨터 비전에서 주로 사용하는 프레임워크로는 OpenCV, TensorFlow, PyTorch 등이 있습니다.

- **OpenCV**: 이미지 처리와 컴퓨터 비전 작업을 위한 오픈소스 라이브러리입니다. 빠른 속도와 다양한 기능을 제공하여 전처리와 간단한 비전 작업에 많이 활용됩니다.
- **TensorFlow**: Google이 개발한 딥러닝 프레임워크입니다. 대규모 배포와 성능 최적화에 강점이 있습니다.
- **PyTorch**: 직관적인 코드 작성과 디버깅이 쉬워 연구와 실험에서 많이 사용하는 딥러닝 프레임워크입니다.

TensorFlow와 PyTorch를 활용하면 합성곱 신경망(CNN), 객체 검출, 이미지 분할 등의 작업을 구현할 수 있습니다.

## 2. 컴퓨터 비전 데이터셋

컴퓨터 비전에서 데이터셋은 모델의 성능을 결정하는 가장 중요한 요소 중 하나입니다. 딥러닝 모델은 대량의 이미지 데이터를 학습해 패턴을 인식하고 일반화하는 능력을 갖추므로 양질의 데이터셋이 필수적입니다.

데이터셋이 다양하고 균형 잡혀 있을수록 모델은 과적합을 피하고 실제 환경에서도 잘 작동할 가능성이 높습니다. ImageNet, COCO, Pascal VOC 같은 대규모 데이터셋은 객체 분류와 검출 모델 학습에 널리 사용되고, MNIST, CIFAR-10, Fashion-MNIST 등은 기초 이미지 분류 실험에 활용됩니다. 의료 영상이나 자율주행처럼 특정 분야에 특화된 데이터셋도 있습니다.

좋은 데이터셋을 구축하려면 이미지 품질, 다양한 환경의 데이터 확보, 정확한 라벨링이 중요합니다. 최근에는 데이터 증강(Augmentation)과 생성 모델(GANs)을 활용해 부족한 데이터를 보완하기도 합니다.

### 주요 데이터셋

#### 1. AI Hub

[AI Hub](https://www.aihub.or.kr/)는 한국의 인공지능 데이터 및 알고리즘 공유 플랫폼입니다. 한국지능정보사회진흥원(NIA)이 운영하며 자연어 처리, 컴퓨터 비전, 음성 인식, 자율주행 등 다양한 AI 분야에서 활용할 수 있는 국내 특화 데이터셋, 학습용 API, 개발 도구 등을 제공합니다. 특히 한국어 기반 데이터가 많아 한국어 AI 모델 개발에 유용합니다.

#### 2. COCO

[COCO(Common Objects in Context)](https://cocodataset.org/#home)는 객체 검출(Object Detection), 이미지 분할(Image Segmentation), 이미지 캡셔닝(Image Captioning) 연구에 널리 쓰이는 대규모 데이터셋입니다. 약 33만 개의 이미지와 150만 개 이상의 객체 주석을 포함합니다. 정밀한 객체 경계 정보, 여러 객체가 포함된 이미지, 객체 클래스·키포인트·캡션 등 풍부한 주석 정보가 특징입니다.

#### 3. Open Images Dataset

[Open Images Dataset](https://storage.googleapis.com/openimages/web/download_v7.html)은 Google이 제공하는 대규모 컴퓨터 비전 데이터셋입니다. 객체 검출, 이미지 분류, 이미지 세그멘테이션 등의 연구를 위해 설계되었습니다. 900만 개 이상의 이미지와 약 600개의 객체 클래스를 제공하며, 바운딩 박스 외에도 관계 주석, 행동 태그, 랜드마크 정보 등을 포함합니다.

#### 4. KITTI

[KITTI 데이터셋](https://www.cvlibs.net/datasets/kitti/)은 자율주행과 컴퓨터 비전 연구를 위한 대표적인 데이터셋입니다. 실제 도로 환경의 고해상도 영상, LiDAR 포인트 클라우드, 스테레오 영상, GPS/IMU 데이터를 제공합니다. 객체 검출, 주행 경로 예측, 깊이 추정, SLAM 등의 연구에 활용됩니다.

## 3. 어노테이션

어노테이션(Annotation)은 머신러닝, 특히 지도학습(Supervised Learning)에서 데이터에 정답인 라벨을 부여하는 과정입니다. 컴퓨터 비전에서는 이미지나 영상 속 객체의 위치, 크기, 카테고리 등을 사람이 직접 지정하거나 자동화 도구를 이용해 태그합니다.

어노테이션 데이터는 딥러닝 모델이 객체를 학습하고 인식하는 데 필수적이며, 데이터 품질은 모델 성능에 직접적인 영향을 줍니다. 최근에는 자동 어노테이션 도구나 크라우드소싱 플랫폼으로 대량의 데이터를 효율적으로 라벨링하는 기술도 발전하고 있습니다.

참고: [어노테이션 관련 자료](https://dev-sites.tistory.com/90)

### 어노테이션 유형

#### 1. 바운딩 박스(Bounding Box)

객체를 감싸는 사각형 형태의 어노테이션으로, 객체 검출에서 가장 널리 사용됩니다. COCO와 PASCAL VOC 데이터셋이 대표적인 예입니다.

#### 2. 폴리곤(Polygon)

불규칙한 형태의 객체를 더 정확하게 표현하기 위해 다각형 형태로 라벨링합니다. 자율주행 차량의 도로 표지판이나 건물 경계선을 표시할 때 사용할 수 있습니다.

#### 3. 세그멘테이션(Segmentation)

픽셀 단위로 객체를 구분합니다. 개별 객체를 구별하는 인스턴스 세그멘테이션(Instance Segmentation)과 같은 의미의 모든 픽셀을 구분하는 의미론적 세그멘테이션(Semantic Segmentation)으로 나뉩니다. COCO와 ADE20K가 대표적인 데이터셋입니다.

#### 4. 키포인트(Keypoint)

얼굴 인식이나 동작 분석 등에 사용합니다. 눈, 코, 손목, 팔꿈치 같은 특정 지점에 점을 찍어 라벨링합니다. OpenPose와 COCO Keypoints가 대표적인 예입니다.

#### 5. 스켈레톤(Skeleton)

여러 키포인트를 선으로 연결하여 인체 구조나 포즈를 표현합니다. 행동 인식이나 스포츠 분석에 활용됩니다.

#### 6. 라인(Line) 및 폴리라인(Polyline)

도로 차선이나 경계선을 선 형태로 라벨링합니다. 자율주행의 도로 차선 감지에 활용됩니다.

#### 7. 3D 바운딩 박스(3D Bounding Box)

객체의 길이, 너비, 높이 정보를 포함하여 3차원 공간에서 라벨링합니다. 자율주행의 LiDAR 데이터 등에 사용됩니다.
