# **1. 주택 임대료 예측 데이터셋**
House Rent Prediction Dataset은 주택 임대료를 예측하기 위한 목적으로 사용되는 데이터셋입니다. 이 데이터셋은 주로 머신러닝 및 데이터 분석 프로젝트에서 사용되며, 주택의 다양한 특성과 위치 정보를 기반으로 임대료를 예측하는 모델을 학습하는 데 활용됩니다.

# **2. 데이터셋 컬럼**
- Posted On: 올라온 날짜
- BHK: 주택에 포함된 침실, 거실, 주방의 총 개수를 의미합니다.
- Rent: 주택(아파트/플랫)의 월 임대료를 나타냅니다.
- Size: 주택(아파트/플랫)의 면적을 평방피트(Square Feet)로 나타냅니다.
- Floor: 주택이 위치한 층수와 건물의 총 층수를 나타냅니다. (예: 2층 중 1층, 5층 중 3층 등)
- Area Type: 주택의 면적이 어떤 방식으로 계산되었는지를 나타냅니다. (예: 전체 면적, 실사용 면적, 건축 면적 등)
- Area Locality: 주택(아파트/플랫)이 위치한 구체적인 지역이나 동네 정보를 나타냅니다.
- City: 주택(아파트/플랫)이 위치한 도시를 나타냅니다.
- Furnishing Status: 주택이 가구가 완비되었는지(Furnished), 부분적으로 갖추어졌는지(Semi-Furnished), 아니면 비어 있는지(Unfurnished)를 나타냅니다.
- Tenant Preferred: 집주인 또는 중개인이 선호하는 임차인 유형을 나타냅니다. (예: 가족, 싱글, 직장인 등)
- Bathroom: 주택에 있는 욕실의 개수를 나타냅니다.
- Point of Contact: 주택(아파트/플랫)에 대한 추가 정보를 얻기 위해 연락해야 할 담당자나 중개인의 정보를 나타냅니다.

### boxplot

Boxplot은 데이터의 중앙값, 사분위수, 이상치 등을 시각적으로 표현하는 통계 그래프입니다. 주로 데이터 분포와 이상치를 빠르게 파악하기 위해 사용됩니다.

1. 중앙값 (Median, Q2): 데이터를 크기 순으로 정렬했을 때 중간에 위치한 값
2. Q1 (제1사분위수, 25%): 하위 25%에 해당하는 값
3. Q3 (제3사분위수, 75%): 상위 25%에 해당하는 값
4. IQR (Interquartile Range, 사분위 범위): Q3 - Q1, IQR은 데이터의 중간 50% 범위를 의미합니다.
5. Minimum: Q1 − 1.5 × IQR 이하에 속하지 않는 가장 작은 값
6. Maximum: Q3 + 1.5 × IQR 이하에 속하지 않는 가장 큰 값
7. 이상치: 일반적인 데이터 분포를 벗어난 값(Lower Bound=Q1−1.5×IQR, Upper Bound=Q3+1.5×IQR)

( IQR 기준으로 이상치를 정의하기 때문에 모든 상황에 완벽하지 않을 수 있습니다)

# **3. PipeLine**
- 여러 작업을 정해진 순서대로 연결하는 객체

```
원본 데이터 → 결측치 처리 → 범주형 데이터 원-핫 인코딩 → 모델 학습

pipe = Pipeline([
    ('단계이름1', 변환기1),
    ('단계이름1', 변환기1),
    ('model', 모델)
])
```

- 각 튜플의 첫번째 값은 단계의 이름이고, 두 번째 값은 실제 작업을 수행할 객체
- 예) ('imputer', SimpleImputer(strategy='median'))

### PipeLine을 사용하는 이유
그니까 머신러닝은 모델만 학습하는게 아님. (원본데이터>결측치처리>스케일링>특성선택>모델학습>예측)이걸 하나씪 매번 적으면 코드도 길어지지만 [순서 실수, 전처리 빼먹음, 테스트데이터 처리 잘못하는거임~~]
```
X_train_imputed = imputer.fit_transform(X_train)
X_train_encoded = encoder.fit_transform(X_train_imputed) // 숫자로 바꿔줌
model.fit(X_train_encoded, y_train)
```

> 위 방식은 학습 데이터와 테스트 데이터에 서로 다른 처리가 적용되거나, 테스트 데이터까지 이용해 전처리 기준을 학습하는 실수를 만들기 쉬움

그래서! Pipline을 사용하면 아래 한줄로 전처리와 학습 연결가능
```
lr_model.fit(X_train, y_train)
pred = lr_model.predict(X_test)
```
> Pipline의 핵심 장점은 단순히 코드가 짧아지는 것이 아닌, 학습과 예측에 동일한 전처리 규칙을 안전하게 적용하는거임

### 숫자형 전처리 Pipeline
```
# 숫자형 컬럼의 결측치를 각 컬럼의 중앙값으로 채움
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])
```

### 범주형 전처리 Pipline
결측치를 가장 자주 등장한 값으로 채움 → 문자열 범주를 원-핫 인코딩으로 숫자화함
```
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
```

### 전체 Pipeline
```
lr_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])
```

### ColumsTransformer
- 숫자형 컬럼과 범주형 컬럼은 같은 방법으로 처리할 수 없음
- 예) 숫자형의 결측치는 중앙값으로 채울 수 있지만, 범주형 데이터 중앙값을 적용할 수 없음
```
preprocessor = ColumsTransformer([
    ('num', numeric_transformer, numeric_features)
    ('cat', categorical_transformer, categorical_features)
])
```

### fit()을 호출할 때 내부에서 일어나는 일
```
lr_model.fit(X_train, y_train)
```
1. X_train의 숫자형 컬럼에서 중앙값을 계산함
2. X_train의 범주형 컬럼에서 최빈값을 계산함
3. X_train에 존재하는 범주 목록을 학습함
4. 전처리된 데이터를 하나의 숫자 행렬로 만듦
5. 그 결과로 회귀 모델을 학습함

> 중앙값, 최빈값, 범주 목록을 오직 X_train에서만 학습한다는것이 중요!

### predict()를 호출할 때 내부에서 일어나는 일
```
pred = lr_model.predict(X_test)
```
1. 학습 때 ㅈ장한 숫자형 중앙값을 사용함
2. 학습 때 저장한 범주형 최빈값을 사용함
3. 학습 때 원-핫 인코딩 규칙을 사용함
4. 전처리 결과를 이미 학습된 모델에 넣어 예측함

> 예측할 때 새로운 중앙값이나 범주 목록을 다시 계산하지 않음

> fit()은 기준을 학습하고 저장하며, predict()는 저장된 기준을 그대로 사용함

### 데이터 누수
- 전체 데이터에서 먼저 전처리 기준을 학습한 후 데이터를 나누는 것
```
X_all_encoded = encoder.fit_transform(X)
X_train, X_test = train_test_split(X_all_encoded)
```

> 위 경우 이코더가 테스트 데이터의 범주 정보까지 미리 보게 됨, 이를 데이터 누수라고 함

### handel_unknown='ignore'
- 학습 데이터는 없던 데이터가 테스트 데이터에 등장할 수 있음
- handel_unknown='ignore' 설정이 없으면 새로운 범주를 만났을 때 오류가 발생할 수 있음
- handel_unknown='ignore'를 사용하면 학습때 없던 범주는 해당 원-핫 컬럼들을 모두 0으로 처리


````python
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
````

````python
onehot = OneHotEncoder(handle_unknown='ignore')

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median'))
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', onehot)
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='drop'
)

lr_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

lr_model;
````

````python
# 1. X_train에서 결측치 대체 기준을 학습
# 2. X_train에서 원-핫 인코딩 범주를 학습
# 3. 전처리된 X_train으로 LinearRegression을 학습
lr_model.fit(X_train, y_train);
````

````python
# 학습 때 저장한 전처리 기준을 X_test에 적용한 뒤 예측
pred = lr_model.predict(X_test)
pred
````

````python
# Pipline 내부에 등록된 단계 확인
print(lr_model.named_steps)
````

````python
# 전처리 후 만들어진 최종 특성 이름 확인
feature_names = lr_model.named_steps['preprocessor'].get_feature_names_out()
print('전처리 후 전체 특성 수: ', len(feature_names))
print('앞부분 특성 이름: ', feature_names[:20])
````

사이킷런의 LinearRegression은 최소 제곱법(Ordinary Least Squares, OLS)을 기반으로 학습하며, 입력 데이터와 실제 목표값 사이의 잔차(Residual)의 제곱합을 최소화하는 방식으로 최적의 가중치(Weight)와 절편(Bias)를 구합니다. 내부적으로는 수치적으로 더 안정적인 특이값 분해(Singular Value Decomposition, SVD)를 사용하여 계산됩니다. 이 방식은 데이터셋이 비교적 작고 특성 수가 적당할 때 효율적입니다.

### ※ 최소 제곱법

최소 제곱법(Least Squares Method)은 주어진 데이터 포인트들과 예측 모델(주로 직선) 사이의 오차(Residuals)의 제곱합(Sum of Squared Errors, SSE)을 최소화하여 최적의 예측 모델을 찾아내는 통계적 방법입니다. 이 방법은 선형 회귀 분석에서 가장 널리 사용되며, 데이터 포인트들이 회귀선(Regression Line)에 최대한 가깝도록 조정합니다. 수학적으로는 오차의 제곱합을 최소화하는 기울기(w)와 절편(b)을 계산하여 모델을 최적화합니다.

> 제곱합을 더하는이유가~~ 0에 가까워지는걸 막으려고(+, - 둘 다 나타날테니까)

> 그래서 오차가 젤 적은 기울기를 찾아서(선) 머 해주는거임

- 1 - 잔차 제곱합 / 전체 제곱합 (R2계산법)

### 로그 변환으로 임대료의 큰 값 영향 완화
- 주택 임대료는 꼬리가 긴 분포를 보이고 있음
- 일부 매우 높은 임대료는 제곱 오차를 크게 만들기 때문에 선형 회귀가 고가 주택에 지나치게 영향을 받을 수 있음
- np.log1p()로 목표값을 변환하면 큰 값 사이의 절대적 차이가 압축됨. 따라서 고가 임대료의 영향이 줄어들 수 있지만, 로그 변환이 항상 성능을 개선하거나 데이터를 정규 분포로 만드는 것은 아님(검증 지표로 확인해야 함)

### 정규 분포
- 정규 분포는 데이터가 평균을 중심으로 좌우 대칭을 이루며 종 모양으로 퍼져 있는 통계적 분포
- 대부분의 값은 평균 근처에 몰려 있고, 평균에서 멀어질수록 값의 빈도가 급격히 감소
- 시험 점수, 사람들의 키와 같은 많은 시제 데이터에서 흔하게 나타남

### 데이터 변환
- 원본 목표값: 실제 임대료 단위의 값
- 로그 변환 목표값: 큰 값의 간격을 압축한 값
```
    # log1p(x)는 log(1+x)를 계산하므로 0이 포함되어도 안전
    y_train_log = np.log1p(y_train)
    # 모델은 로그 단위의 임대료를 에측하므로, 최종 평가 전에는 원래 단위로 되돌려야 함
    y_pred = np.expm1(y_pred_log)
```
> 로그 변환은 값의 순서를 유지하지만, 값 사이의 절대적 거리와 관계는 달라짐. 따라서 "관계가 전혀 변하지 않는다"기보다는 "큰 값의 영향이 압축된다"고 이해하는 것이 정확

````python
# 목표값만 로그 변환하여 학습
y_train_log = np.log1p(y_train)
lr_model.fit(X_train, y_train_log)
y_pred_log = lr_model.predict(X_test)

# 로그 단위 에측을 원래 임대료 단위로 복원
log_pred = np.expm1(y_pred_log)

# 임대료가 음수가 되는 비현실적인 예측은 0으로 제한
log_pred = np.clip(log_pred, a_min=0, a_max=None)

````

````python
mae = mean_absolute_error(y_test, log_pred)
rmse = root_mean_squared_error(y_test, log_pred)
mse = mean_squared_error(y_test, log_pred)
r2 = r2_score(y_test, log_pred)

print(f'MAE: {mae:,.2f}')
print(f'RMSE: {rmse:,.2f}')
print(f'MSE: {mse:,.2f}')
print(f'R2: {r2:,.2f}')
````
